```python
students = []

def add_student():
    roll_no = input("Enter Roll Number: ")

    for student in students:
        if student["roll_no"][0] == roll_no:
            print("Student already exists!")
            return

    name = input("Enter Name: ")
    department = input("Enter Department: ")
    reg_no = input("Enter Registration Number: ")
    dob = input("Enter Date of Birth: ")

    subjects = input("Enter Subjects (comma separated): ").split(",")
    marks = list(map(int, input("Enter Marks (comma separated): ").split(",")))

    attendance = float(input("Enter Attendance Percentage: "))
    email = input("Enter Email: ")

    student = {
        "roll_no": (roll_no,),
        "reg_no": (reg_no,),
        "dob": (dob,),
        "name": name,
        "department": department,
        "subjects": subjects,
        "marks": marks,
        "attendance": attendance,
        "email": email
    }

    students.append(student)
    print("Student Added Successfully!")


def search_student():
    roll_no = input("Enter Roll Number to Search: ")

    for student in students:
        if student["roll_no"][0] == roll_no:
            print(student)
            return

    print("Student Not Found")


def update_student():
    roll_no = input("Enter Roll Number to Update: ")

    for student in students:
        if student["roll_no"][0] == roll_no:
            student["name"] = input("Enter New Name: ")
            print("Record Updated Successfully")
            return

    print("Student Not Found")


def delete_student():
    roll_no = input("Enter Roll Number to Delete: ")

    for student in students:
        if student["roll_no"][0] == roll_no:
            students.remove(student)
            print("Record Deleted Successfully")
            return

    print("Student Not Found")


def display_students():
    if len(students) == 0:
        print("No Records Found")
    else:
        for student in students:
            print(student)


def calculate_average():
    roll_no = input("Enter Roll Number: ")

    for student in students:
        if student["roll_no"][0] == roll_no:
            avg = sum(student["marks"]) / len(student["marks"])
            print("Average Marks =", avg)
            return

    print("Student Not Found")


def highest_scorer():
    if len(students) == 0:
        print("No Records Found")
        return

    topper = max(students, key=lambda x: sum(x["marks"]))
    print("Highest Scorer:", topper["name"])


def list_by_department():
    dept = input("Enter Department: ")

    found = False

    for student in students:
        if student["department"].lower() == dept.lower():
            print(student)
            found = True

    if not found:
        print("No Students Found")


def count_students():
    print("Total Students =", len(students))


while True:
    print("\n===== Student Record Management System =====")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display Students")
    print("6. Calculate Average Marks")
    print("7. Find Highest Scorer")
    print("8. List Students by Department")
    print("9. Count Students")
    print("10. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        search_student()
    elif choice == "3":
        update_student()
    elif choice == "4":
        delete_student()
    elif choice == "5":
        display_students()
    elif choice == "6":
        calculate_average()
    elif choice == "7":
        highest_scorer()
    elif choice == "8":
        list_by_department()
    elif choice == "9":
        count_students()
    elif choice == "10":
        print("Thank You")
        break
    else:
        print("Invalid Choice")
```
